```python
import re
import os
import joblib
import numpy as np
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from scipy.sparse import save_npz
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
# Download the NLTK resources (only downloads the first time)
for pkg in ["punkt", "punkt_tab", "stopwords", "wordnet", "omw-1.4"]:
    nltk.download(pkg, quiet=True)

os.makedirs("models", exist_ok=True)
```

```python
df = pd.read_csv("selected_data.csv")
print("Original shape:", df.shape)
```

```python
df.isnull().sum()
```

```python
# Rows without a summary or a label are useless for training
df = df.dropna(subset=["brief_summary", "disease_category"])
```

```python
# Remove rows where the summary is only spaces
df = df[df["brief_summary"].str.strip() != ""]
```

```python
# Remove duplicates
before = len(df)
df = df.drop_duplicates(subset=["brief_summary"], keep="first")
print(f"\nDuplicates removed: {before - len(df)}")
```

```python
df = df.reset_index(drop=True)
print("Shape after missing/duplicate handling:", df.shape)
```

```python
#Text cleaning, normalization, punctuation, special characters
import re

def clean_text(text):
    text = text.lower()                              # lowercase
    text = re.sub(r"http\S+|www\.\S+", " ", text)    # remove URLs
    text = re.sub(r"[^a-z\s]", " ", text)            # keep only letters
                                                     # (removes punctuation,
                                                     # digits, symbols)
    text = re.sub(r"\s+", " ", text).strip()         # remove extra spaces
    return text

df["clean_text"] = df["brief_summary"].apply(clean_text)
```

```python
#Tokenization, stop-word removal and lemmatization
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

def tokenize_and_lemmatize(text):
    tokens = word_tokenize(text)                            # tokenization
    tokens = [t for t in tokens if t not in stop_words]     # remove stop words
    tokens = [t for t in tokens if len(t) > 2]              # remove tiny tokens
    tokens = [lemmatizer.lemmatize(t) for t in tokens]      # lemmatization
    return " ".join(tokens)

print("\nTokenizing and lemmatizing... (this can take 1-2 minutes)")
df["processed_text"] = df["clean_text"].apply(tokenize_and_lemmatize)

```

```python
# Drop rows that became empty after cleaning
df = df[df["processed_text"].str.strip() != ""].reset_index(drop=True)

print("\n--- BEFORE vs AFTER (first row) ---")
print("Original :", df["brief_summary"][0][:250])
print("Processed:", df["processed_text"][0][:250])

df.to_csv("cleaned_data.csv", index=False)
print("\nSaved cleaned_data.csv with shape:", df.shape)
```

```python
# Encode target labels
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(df["disease_category"])

print("\n--- LABEL ENCODING ---")
for i, name in enumerate(label_encoder.classes_):
    print(f"{i} -> {name}")
```

```python
# Train/test split, then TF-IDF vectorization
X_train_text, X_test_text, y_train, y_test = train_test_split(
    df["processed_text"],
    y,
    test_size=0.2,
    random_state=42,
    stratify=y,          # same category proportions in train and test
)
print("\nTrain size:", len(X_train_text), "| Test size:", len(X_test_text))

tfidf = TfidfVectorizer(
    max_features=10000,   # keep the 10,000 most useful words/phrases
    ngram_range=(1, 2),   # single words and 2-word phrases
    min_df=5,             # ignore terms found in fewer than 5 documents
)

X_train = tfidf.fit_transform(X_train_text)   # learn vocabulary + transform
X_test = tfidf.transform(X_test_text)         # only transform

print("TF-IDF train matrix shape:", X_train.shape)
print("TF-IDF test matrix shape :", X_test.shape)
```

```python
# Save everything for the modeling step
joblib.dump(tfidf, "models/tfidf_vectorizer.pkl")
joblib.dump(label_encoder, "models/label_encoder.pkl")
save_npz("X_train.npz", X_train)
save_npz("X_test.npz", X_test)
np.save("y_train.npy", y_train)
np.save("y_test.npy", y_test)

print("\nSaved: vectorizer, label encoder, and train/test matrices.")
```

```python
# 3. Explore relationships between clinical trial summaries and disease categories

# Most distinctive words per category (mean TF-IDF score)
tfidf = TfidfVectorizer(max_features=5000, min_df=5)
X = tfidf.fit_transform(df["processed_text"])
terms = tfidf.get_feature_names_out()

# Use custom_order containing correct category names
custom_order = [
    "breast cancer",
    "type 2 diabetes",
    "covid-19",
    "anxiety",
    "chronic obstructive pulmonary disease",
    "rheumatoid arthritis",
    "glaucoma",
    "sickle cell anemia"
]

mean_tfidf = {}
distinct_rows = []
for cat in custom_order:
    idx = (df["disease_category"] == cat).values
    scores = X[idx].mean(axis=0).A1
    mean_tfidf[cat] = scores
    top_idx = scores.argsort()[::-1][:10]
    distinct_rows.append({"disease_category": cat,
                          "distinctive_words": ", ".join(terms[top_idx])})

pd.set_option("display.max_colwidth", None)
display(pd.DataFrame(distinct_rows))

# Heatmap: how strongly each category uses its signature words
tfidf_df = pd.DataFrame(mean_tfidf, index=terms)
signature = sorted({t for cat in custom_order for t in tfidf_df[cat].nlargest(5).index})
heat = tfidf_df.loc[signature].T

fig = px.imshow(
    heat, aspect="auto", color_continuous_scale="viridis",
    title="Signature Words vs Disease Category (mean TF-IDF)",
    template=TEMPLATE, height=520,
)
fig.update_layout(xaxis_title="Word", yaxis_title="")
fig.show()
```

```python
#Step 4: NLP-Based Disease Classification

# %%
import re
import time

import joblib
import nltk
import numpy as np
import pandas as pd
import plotly.express as px
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from sklearn.base import clone
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, classification_report,
                             confusion_matrix, f1_score, precision_score,
                             recall_score)
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC

try:
    from IPython.display import display
except ImportError:
    display = print

for pkg in ["punkt", "punkt_tab", "stopwords", "wordnet", "omw-1.4"]:
    nltk.download(pkg, quiet=True)

TEMPLATE = "plotly_white"
```

```python
# ## 1 & 2. NLP techniques and TF-IDF features
# The text was cleaned, tokenized, stop words removed and lemmatized in Step 2
# (`processed_text`). Here we split the data and convert the text into numbers
# with an improved TF-IDF vectorizer.

# %%
df = pd.read_csv("cleaned_data.csv")
label_encoder = joblib.load("models/label_encoder.pkl")
class_names = label_encoder.classes_
y = label_encoder.transform(df["disease_category"])

# Same split as Step 2 (80/20, stratified, random_state=42)
X_train_text, X_test_text, y_train, y_test = train_test_split(
    df["processed_text"], y, test_size=0.2, random_state=42, stratify=y)

# Improved TF-IDF settings
#   max_features=50000 : keep more words/phrases (was 10,000)
#   ngram_range=(1,2)  : single words and two-word phrases
#   min_df=2           : ignore terms found in only 1 document (was 5)
#   sublinear_tf=True  : dampen very frequent words (1 + log(tf))
tfidf = TfidfVectorizer(max_features=50000, ngram_range=(1, 2),
                        min_df=2, sublinear_tf=True)
X_train = tfidf.fit_transform(X_train_text)   # learn vocabulary on train only
X_test = tfidf.transform(X_test_text)

print("Training matrix:", X_train.shape)
print("Testing matrix :", X_test.shape)
print("Classes        :", list(class_names))

features = tfidf.get_feature_names_out()
print("\nSample TF-IDF features:", list(features[:15]))

# Which words get the highest average weight?
mean_weight = np.asarray(X_train.mean(axis=0)).ravel()
top = pd.DataFrame({"term": features, "mean_tfidf": mean_weight}).nlargest(20, "mean_tfidf")
fig = px.bar(top, x="mean_tfidf", y="term", orientation="h",
             title="Top 20 TF-IDF Terms (average weight in training data)",
             template=TEMPLATE, height=550)
fig.update_layout(yaxis={"categoryorder": "total ascending"}, yaxis_title="", xaxis_title="Mean TF-IDF")
fig.show()
```

```python
 ## 3. Train machine learning models

# %%
models = {
    "Logistic Regression": LogisticRegression(C=10, max_iter=2000,
                                              class_weight="balanced", random_state=42),
    "Linear SVM": LinearSVC(C=1, class_weight="balanced", random_state=42),
    "Naive Bayes": MultinomialNB(),
    "Random Forest": RandomForestClassifier(n_estimators=100, class_weight="balanced",
                                            n_jobs=-1, random_state=42),
}

results, predictions, trained = [], {}, {}

for name, model in models.items():
    print(f"Training {name} ...", end=" ")
    start = time.time()
    model.fit(X_train, y_train)
    seconds = time.time() - start

    y_pred = model.predict(X_test)
    predictions[name] = y_pred
    trained[name] = model

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision (macro)": precision_score(y_test, y_pred, average="macro"),
        "Recall (macro)": recall_score(y_test, y_pred, average="macro"),
        "F1 (macro)": f1_score(y_test, y_pred, average="macro"),
        "F1 (weighted)": f1_score(y_test, y_pred, average="weighted"),
        "Train time (s)": round(seconds, 1),
    })
    print(f"done in {seconds:.1f}s")
```

```python
## 4 & 5. Classify the summaries and compare the algorithms

# %%
results_df = pd.DataFrame(results).sort_values("Accuracy", ascending=False).reset_index(drop=True)
pd.set_option("display.width", 200)
pd.set_option("display.max_columns", None)
display(results_df.round(4))

# Final model chosen manually: Logistic Regression (see the note in the next cell)
best_name = "Logistic Regression"
print(f"\nSelected model: {best_name}")
```

### Final model: Logistic Regression

Linear SVM scored slightly higher (accuracy 95.5%), but Logistic Regression was very close (accuracy 95.2%).
The difference is less than 0.5%.

I selected **Logistic Regression** because:
- It gives a probability for each disease, which I use as the confidence score.
- It is simple and easy to explain.
- It works well for TF-IDF text classification.

```python
## 6. Visualize model performance

# %%
metrics = ["Accuracy", "Precision (macro)", "Recall (macro)", "F1 (macro)"]
long = results_df.melt(id_vars="Model", value_vars=metrics, var_name="Metric", value_name="Score")

fig = px.bar(long, x="Model", y="Score", color="Metric", barmode="group",
             text=long["Score"].round(3), title="Model Comparison",
             template=TEMPLATE, height=520)
fig.update_yaxes(range=[0.5, 1.0])
fig.update_traces(textposition="outside")
fig.show()

# %%
fig = px.bar(results_df, x="Model", y="Train time (s)", text="Train time (s)",
             title="Training Time per Model", template=TEMPLATE)
fig.show()
```

```python
# Classification report of the best model
y_best = predictions[best_name]
report = classification_report(y_test, y_best, target_names=class_names, output_dict=True)
report_df = pd.DataFrame(report).T.round(3)
print(f"Accuracy: {accuracy_score(y_test, y_best):.3f}")
print(f"Classification report - {best_name}")
display(report_df)
```

```python
# Step 5: Disease Prediction System

# %%
import re

import joblib
import nltk
import numpy as np
import pandas as pd
import plotly.express as px
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from sklearn.base import clone
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import (accuracy_score, classification_report,
                             confusion_matrix, f1_score, precision_score,
                             recall_score)
from sklearn.model_selection import StratifiedKFold, cross_validate, train_test_split
from sklearn.pipeline import make_pipeline

try:
    from IPython.display import display
except ImportError:
    display = print

for pkg in ["punkt", "punkt_tab", "stopwords", "wordnet", "omw-1.4"]:
    nltk.download(pkg, quiet=True)

TEMPLATE = "plotly_white"
DATA_PATH = "cleaned_data.csv"        # use "data/cleaned_data.csv" if needed
```

```python
# 1. Load the trained system (TF-IDF vectorizer + model + labels)

tfidf = joblib.load("models/tfidf_vectorizer.pkl")
label_encoder = joblib.load("models/label_encoder.pkl")
best_model = joblib.load("models/best_model.pkl")
class_names = label_encoder.classes_

print("Model type      :", type(best_model).__name__)
print("TF-IDF features :", len(tfidf.get_feature_names_out()))
print("Categories      :", list(class_names))
```

```python
# ## 2. Use the Brief Summary text data (same train/test split as Steps 2 and 4)

# %%
df = pd.read_csv(DATA_PATH)
y = label_encoder.transform(df["disease_category"])

X_train_text, X_test_text, y_train, y_test = train_test_split(
    df["processed_text"], y, test_size=0.2, random_state=42, stratify=y)

# Re-apply TF-IDF using the current in-memory tfidf (50,000 features)
# and make sure it is dumped to pkl to avoid mismatch in the future
joblib.dump(tfidf, "models/tfidf_vectorizer.pkl")

X_train = tfidf.transform(X_train_text)
X_test = tfidf.transform(X_test_text)
print("Train:", X_train.shape, "| Test:", X_test.shape)

# Verify that they match up correctly
n_model_features = best_model.coef_.shape[1] if hasattr(best_model, "coef_") else X_test.shape[1]
assert n_model_features == X_test.shape[1], f"tfidf_vectorizer ({X_test.shape[1]}) and best_model ({n_model_features}) do not match."
```

```python
rows = []
for name, m in trained.items():
    p_train, p_test = m.predict(X_train), m.predict(X_test)
    rows.append({
        "Model": name,
        "Train accuracy": accuracy_score(y_train, p_train),
        "Test accuracy": accuracy_score(y_test, p_test),
        "Gap (train - test)": accuracy_score(y_train, p_train) - accuracy_score(y_test, p_test),
        "Train F1 (macro)": f1_score(y_train, p_train, average="macro"),
        "Test F1 (macro)": f1_score(y_test, p_test, average="macro"),
    })
fit_df = pd.DataFrame(rows).sort_values("Test accuracy", ascending=False).round(4)
display(fit_df)

fig = px.bar(fit_df.melt(id_vars="Model", value_vars=["Train accuracy", "Test accuracy"]),
             x="Model", y="value", color="variable", barmode="group",
             text_auto=".3f", title="Train vs Test Accuracy", template=TEMPLATE)
fig.update_yaxes(range=[0.8, 1.0])
fig.update_layout(yaxis_title="Accuracy", legend_title="")
fig.show()
```

```python
 #3. Add confidence scores
# Linear SVM only gives a decision score, not a probability. We calibrate it so
# the app can show a real confidence (%) for each prediction.

# %%
if hasattr(best_model, "predict_proba"):
    final_model = best_model                       # already gives probabilities
else:
    final_model = CalibratedClassifierCV(clone(best_model), cv=5)
    final_model.fit(X_train, y_train)

acc_before = accuracy_score(y_test, best_model.predict(X_test))
acc_after = accuracy_score(y_test, final_model.predict(X_test))
print(f"Accuracy of original model  : {acc_before:.4f}")
print(f"Accuracy with confidence    : {acc_after:.4f}")
```

```python
## 4. Predict disease categories from medical text input

# %%
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    """Same cleaning pipeline as Step 2."""
    text = text.lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)
    text = re.sub(r"[^a-z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    tokens = word_tokenize(text)
    tokens = [t for t in tokens if t not in stop_words and len(t) > 2]
    return " ".join(lemmatizer.lemmatize(t) for t in tokens)

def predict_disease(summary, top_n=3):
    """Return the predicted category, confidence and the top-N candidates."""
    vec = tfidf.transform([preprocess_text(summary)])
    proba = final_model.predict_proba(vec)[0]
    order = proba.argsort()[::-1][:top_n]
    return {
        "category": class_names[order[0]],
        "confidence": float(proba[order[0]]),
        "top": [(class_names[i], float(proba[i])) for i in order],
    }

samples = [
    "This study evaluates a new eye drop to lower intraocular pressure in patients with open angle disease.",
    "A randomized trial of cognitive behavioral therapy to reduce worry and nervousness in college students.",
    "We test whether a vaccine booster reduces severe infection caused by the SARS-CoV-2 virus.",
    "Hydroxyurea dosing to reduce pain crises in children with abnormal hemoglobin.",
    "Effect of a bronchodilator inhaler on breathlessness and exacerbations in smokers with airflow limitation.",
    "Depression and anxiety screening in women after breast surgery.",   # mixed topic
]

rows = []
for s in samples:
    r = predict_disease(s)
    rows.append({
        "Summary": s[:70] + "...",
        "Predicted": r["category"],
        "Confidence": f"{r['confidence']:.1%}",
        "2nd choice": f"{r['top'][1][0]} ({r['top'][1][1]:.1%})",
    })
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", 250)
display(pd.DataFrame(rows))

# %%
# Probability chart for one summary
result = predict_disease(samples[5], top_n=len(class_names))
chart = pd.DataFrame(result["top"], columns=["disease_category", "probability"])
fig = px.bar(chart, x="probability", y="disease_category", orientation="h",
             text=chart["probability"].map("{:.1%}".format),
             title="Prediction Confidence for One Summary", template=TEMPLATE, height=420)
fig.update_layout(yaxis={"categoryorder": "total ascending"}, yaxis_title="", xaxis_title="Probability")
fig.show()
```

```python
 ## 5. Validate prediction accuracy and model performance

# %%
y_pred = final_model.predict(X_test)

scores = pd.DataFrame({
    "Metric": ["Accuracy", "Precision (macro)", "Recall (macro)", "F1 (macro)", "F1 (weighted)"],
    "Score": [accuracy_score(y_test, y_pred),
              precision_score(y_test, y_pred, average="macro"),
              recall_score(y_test, y_pred, average="macro"),
              f1_score(y_test, y_pred, average="macro"),
              f1_score(y_test, y_pred, average="weighted")],
}).round(4)
display(scores)

fig = px.bar(scores, x="Metric", y="Score", text="Score", title="Test Set Performance",
             template=TEMPLATE, height=420)
fig.update_yaxes(range=[0.5, 1.0])
fig.update_traces(textposition="outside")
fig.show()

# %%
report = classification_report(y_test, y_pred, target_names=class_names, output_dict=True)
display(pd.DataFrame(report).T.round(3))

# %%
cm = confusion_matrix(y_test, y_pred)
cm_pct = (cm / cm.sum(axis=1, keepdims=True) * 100).round(1)
fig = px.imshow(cm_pct, x=list(class_names), y=list(class_names), text_auto=True,
                color_continuous_scale="Blues", aspect="auto",
                title="Confusion Matrix (% of true category)", template=TEMPLATE, height=650)
fig.update_layout(xaxis_title="Predicted", yaxis_title="Actual")
fig.show()
```

```python
# %%
# 5-fold cross-validation: a more reliable estimate than a single split.
# A fresh TF-IDF is fitted inside every fold, so there is no leakage.
pipe = make_pipeline(clone(tfidf), clone(best_model))
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_res = cross_validate(pipe, df["processed_text"], y, cv=cv,
                        scoring=["accuracy", "f1_macro"])

cv_df = pd.DataFrame({
    "Fold": [f"Fold {i}" for i in range(1, 6)],
    "Accuracy": cv_res["test_accuracy"],
    "F1 (macro)": cv_res["test_f1_macro"],
}).round(4)
display(cv_df)
print(f"Accuracy   : {cv_df['Accuracy'].mean():.4f} +/- {cv_df['Accuracy'].std():.4f}")
print(f"F1 (macro) : {cv_df['F1 (macro)'].mean():.4f} +/- {cv_df['F1 (macro)'].std():.4f}")

fig = px.bar(cv_df.melt(id_vars="Fold"), x="Fold", y="value", color="variable", barmode="group",
             text_auto=".3f", title="5-Fold Cross-Validation", template=TEMPLATE, height=450)
fig.update_yaxes(range=[0.5, 1.0])
fig.update_layout(yaxis_title="Score", legend_title="")
fig.show()
```

```python
# Is the confidence score trustworthy? Accuracy should rise with confidence.
proba = final_model.predict_proba(X_test)
conf = proba.max(axis=1)
correct = (proba.argmax(axis=1) == y_test)

bins = [0, 0.5, 0.7, 0.9, 1.0001]
labels = ["< 50%", "50-70%", "70-90%", "90-100%"]
bucket = pd.cut(conf, bins=bins, labels=labels, right=False)
rel = (pd.DataFrame({"Confidence": bucket, "Correct": correct})
         .groupby("Confidence", observed=True)["Correct"].agg(["mean", "count"]).reset_index())
rel.columns = ["Confidence level", "Accuracy", "Predictions"]
display(rel.round(4))

fig = px.bar(rel, x="Confidence level", y="Accuracy", text=rel["Accuracy"].map("{:.1%}".format),
             hover_data=["Predictions"], title="Prediction Accuracy by Confidence Level",
             template=TEMPLATE, height=420)
fig.update_yaxes(range=[0, 1.05])
fig.show()
```
